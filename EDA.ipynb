## Univariate Analysis

```python
for col in numeric_col:
  plt.figure(figsize=(12,4))

  plt.subplot(1,3,1)
  sns.histplot(df[col],kde=True)
  plt.title(f"Histogram of {col}")

  plt.subplot(1,3,2)
  sns.boxplot(df[col])
  plt.title(f"Boxplot of {col}")

  plt.subplot(1,3,3)
  sns.kdeplot(df[col],fill=True , warn_singular=False)
  plt.title(f"KDE Plot of {col}")

  plt.tight_layout()
  plt.show()
```

# Bivariate Analysis

## 1.Numeric <--> Numeric

```python
df = pd.read_csv("DataSet/car_data.csv")
```

```python
df
```

```python
df.columns
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
sns.scatterplot(x="Year" , y = "Selling_Price" , data=df)
plt.title("Scatter plot for Year VS Selling Price")

plt.subplot(1,3,2)
sns.regplot(x = "Year" , y = "Selling_Price" , data=df)
plt.title("Regression plot for Year VS Selling Price")

plt.subplot(1,3,3)
sns.lineplot(x="Year" , y = "Selling_Price" , data=df)
plt.title("Line plot for Year VS Selling Price")

plt.tight_layout()
plt.show()
```

### 2 . Numeric <--> category

```python
plt.figure(figsize = (13,4))

plt.subplot(1,3,1)
sns.boxplot(x = "Fuel_Type" , y="Present_Price" , data=df)
plt.title("Present_price VS Fuel type")

plt.subplot(1,3,2)
sns.violinplot(x = "Fuel_Type" , y="Present_Price" , data=df)
plt.title("Present_price VS Fuel Type")

plt.subplot(1,3,3)
sns.barplot(x = "Fuel_Type" , y="Present_Price" , data=df)
plt.title("Present_price VS Fuel Type")

plt.tight_layout()
plt.show()
```

### 3.Categorical <--> Categorical

```python
plt.figure(figsize = (13,4))

plt.subplot(1,3,1)
sns.countplot(x = "Fuel_Type" , hue="Seller_Type" , data=df)
plt.title("Present_price VS Fuel type")

plt.subplot(1,3,2)
sns.violinplot(x = "Fuel_Type" , y="Seller_Type" , data=df)
plt.title("Present_price VS Fuel Type")

plt.subplot(1,3,3)
sns.heatmap(df.corr(numeric_only = True) , annot=True)
plt.title("Present_price VS Fuel Type")

plt.tight_layout()
plt.show()
```
